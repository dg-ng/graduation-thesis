## Figure 3.2 — Topic distribution in the text dataset
Horizontal bar chart of 16 categories (source: `data/master_data.csv`).  
Bars sharing a colour belong to an overlapping semantic cluster.

```python
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
```

```python
df = pd.read_csv('../data/master_data.csv')
counts = df['category'].value_counts()
print(f'Categories: {len(counts)}  |  Total rows: {len(df)}')
print(counts.to_string())
```

```python
# ── Semantic clusters (overlapping pairs / groups) ────────────────────────────
# Each cluster maps to one categorical colour slot from the validated palette
# (palette.md: blue #2a78d6 · orange #eb6834 · aqua #1baf7a · yellow #eda100
#              magenta #e87ba4 · green #008300 · violet #6250d6 · red #e34948)

CLUSTERS = {
    'News / Politics':     (['POLITICS', 'NEWS'], '#2a78d6'),        # slot 1 blue
    'Entertainment':       (['ENTERTAINMENT', 'COMEDY',
                             'MEDIA', 'ARTS & CULTURE'], '#eb6834'), # slot 2 orange
    'Science / Nature':    (['SCIENCE', 'HEALTH',
                             'ENVIRONMENT'], '#1baf7a'),             # slot 3 aqua
    'Business / Tech':     (['BUSINESS', 'TECH'], '#eda100'),        # slot 4 yellow
    'Social':              (['WOMEN', 'EDUCATION'], '#e87ba4'),      # slot 5 magenta
    'Sports':              (['SPORTS'], '#008300'),                  # slot 6 green
    'Religion':            (['RELIGION'], '#6250d6'),                # slot 7 violet
    'Crime':               (['CRIME'], '#e34948'),                   # slot 8 red
}

cat_to_cluster  = {cat: name for name, (cats, _) in CLUSTERS.items() for cat in cats}
cat_to_color    = {cat: color for _, (cats, color) in CLUSTERS.items() for cat in cats}
cluster_order   = list(CLUSTERS.keys())
```

```python
# ── Sort bars: by cluster order first, then alphabetically within cluster ─────
records = [
    {'category': cat, 'count': cnt,
     'cluster': cat_to_cluster[cat], 'color': cat_to_color[cat]}
    for cat, cnt in counts.items()
]
plot_df = (
    pd.DataFrame(records)
      .assign(cluster_rank=lambda d: d['cluster'].map(
          {name: i for i, name in enumerate(cluster_order)}))
      .sort_values(['cluster_rank', 'category'], ascending=[True, True])
      .reset_index(drop=True)
)

# ── Chart constants ───────────────────────────────────────────────────────────
SURFACE   = '#fcfcfb'
INK       = '#0b0b0b'
INK_MED   = '#52514e'
INK_MUTED = '#898781'
GRID      = '#e1e0d9'
BAR_H     = 0.55

n = len(plot_df)
fig_h = max(5, n * 0.42)

fig, ax = plt.subplots(figsize=(8, fig_h), facecolor=SURFACE)
ax.set_facecolor(SURFACE)

# ── Bars ──────────────────────────────────────────────────────────────────────
for i, row in plot_df.iterrows():
    ax.barh(i, row['count'], height=BAR_H, color=row['color'], linewidth=0)

# ── Gridlines (recessive hairline) ───────────────────────────────────────────
ax.set_axisbelow(True)
ax.xaxis.grid(True, color=GRID, linewidth=1, linestyle='solid')
ax.yaxis.grid(False)

# ── Axes ticks & labels ───────────────────────────────────────────────────────
ax.set_yticks(range(n))
ax.set_yticklabels(plot_df['category'].str.title(), fontsize=10, color=INK)
ax.set_xlim(0, counts.max() * 1.18)
ax.set_xticks([0, 100, 200, 300])
ax.tick_params(axis='x', colors=INK_MUTED, labelsize=9)
ax.tick_params(axis='y', length=0)

for spine in ['top', 'right', 'bottom']:
    ax.spines[spine].set_visible(False)
ax.spines['left'].set_color(GRID)

# ── "n = 300" annotation ──────────────────────────────────────────────────────
ax.axvline(300, color=GRID, linewidth=1, linestyle='solid', zorder=0)
ax.text(302, 0, 'n = 300\nper category', fontsize=8, color=INK_MUTED, va='top')

# ── Legend ────────────────────────────────────────────────────────────────────
legend_handles = [
    mpatches.Patch(facecolor=color, label=name, linewidth=0)
    for name, (_, color) in CLUSTERS.items()
]
ax.legend(
    handles=legend_handles,
    title='Cluster',
    title_fontsize=8,
    fontsize=8,
    loc='lower right',
    frameon=True,
    framealpha=0.95,
    edgecolor=GRID,
    labelcolor=INK,
)

ax.invert_yaxis()

# ── Title & subtitle (figure-level, above the axes) ──────────────────────────
plt.tight_layout()
fig.subplots_adjust(top=0.88)
fig.text(0.02, 0.96, 'Topic distribution — text dataset',
         fontsize=12, fontweight='bold', color=INK, va='bottom')
fig.text(0.02, 0.92,
         'Colour encodes semantic cluster; bars sharing a colour may overlap in labelling tasks.',
         fontsize=8.5, color=INK_MED, va='bottom')

plt.savefig('../docs/fig3_2_category_distribution.png',
            dpi=150, bbox_inches='tight', facecolor=SURFACE)
plt.show()
print('Saved → docs/fig3_2_category_distribution.png')
```
